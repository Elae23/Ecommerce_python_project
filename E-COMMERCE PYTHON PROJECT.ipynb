```python
# Select only your date columns and show their data types
date_columns_to_check = [
    'order_purchase_timestamp',
    'order_approved_at',
    'order_delivered_carrier_date',
    'order_delivered_customer_date',
    'order_estimated_delivery_date',
    'shipping_limit_date',
    'review_creation_date',
    'review_answer_timestamp'
]

existing_columns = [
    col for col in date_columns_to_check
    if col in Ecommerce_clean.columns
]

print(Ecommerce_clean[existing_columns].dtypes)
```

```python
num_cols = Ecommerce_clean.select_dtypes(include=['number']).columns

for col in num_cols:

    Q1 = Ecommerce_clean[col].quantile(0.25)

    Q3 = Ecommerce_clean[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR

    upper = Q3 + 1.5 * IQR

    outliers = Ecommerce_clean[
        (Ecommerce_clean[col] < lower) |
        (Ecommerce_clean[col] > upper)
    ]

    print(col, len(outliers))
```

* Discussion of Outliers

 Outliers were identified using the Interquartile Range (IQR) method, where observations below Q1−1.5×IQR or above Q3+1.5×IQR were classified as outliers. The analysis showed varying numbers of outliers across the numerical variables.

 The variables customer_zip_code_prefix and geolocation_zip_code_prefix recorded no outliers. This suggests that the zip code prefixes are relatively consistent within the dataset. It is also worth noting that zip code prefixes are identifiers rather than continuous measurements, so they are not typically interpreted in terms of outliers.

 Several variables showed relatively small numbers of outliers. For example, product_name_lenght (1,300), product_photos_qty (3,205), product_width_cm (2,754), product_length_cm (3,822), and payment_sequential (5,141). These likely reflect products with unusually long names, a higher number of photos, larger dimensions, or orders with multiple payment attempts. Such observations are plausible in an e-commerce setting.

 Other variables exhibited a moderate to high number of outliers. These include price (8,980), freight_value (12,790), payment_value (9,638), payment_installments (8,153), product_description_lenght (7,664), product_height_cm (8,101), delivery_days (5,812), and product_weight_g (16,696). These outliers may represent expensive products, unusually high shipping costs, heavier or larger products, extended product descriptions, or deliveries that took significantly longer than the majority of orders.

The largest number of outliers was observed in review_score (19,587), followed by geolocation_lat (19,415) and seller_zip_code_prefix (18,296). For review_score, this result is expected because the variable is measured on a discrete scale of 1 to 5, making the IQR method less appropriate for identifying unusual values. Similarly, geolocation_lat and seller_zip_code_prefix represent geographical coordinates and identifiers rather than continuous business measurements, so the IQR method may incorrectly classify many valid observations as outliers.

 Overall, the detected outliers are largely attributable to the natural variability of e-commerce transactions, including differences in product prices, product characteristics, shipping costs, customer locations, and delivery performance. Since these observations likely represent legitimate business activities rather than data entry errors, they were retained in the dataset to preserve the integrity and representativeness of the analysis.

```python
#Question 4 analyze one and returning customers
#Step 1: Count orders per customer

customer_orders = (
    Ecommerce.groupby('customer_unique_id')['order_id']
    .nunique()
    .reset_index(name='Number_of_Orders')
)

customer_orders.head()

#Step 2: Count one-time and returning customers
one_time = (customer_orders['Number_of_Orders'] == 1).sum()
returning = (customer_orders['Number_of_Orders'] > 1).sum()

print(f"One-time customers: {one_time}")
print(f"Returning customers: {returning}")

#Step 3: Calculate percentages
total_customers = len(customer_orders)

summary = pd.DataFrame({
    'Customer Type': ['One-time', 'Returning'],
    'Count': [one_time, returning]
})

summary['Percentage (%)'] = (
    summary['Count'] / total_customers * 100
).round(2)

summary
```

Interpretation of Customer Purchase Frequency

The analysis of customer purchase frequency indicates that the vast majority of customers made only a single purchase on the Olist e-commerce platform. Out of 96,096 customers, 93,099 (96.88%) were one-time customers, while only 2,997 (3.12%) were returning customers who made more than one purchase.

This finding suggests that customer retention on the platform was relatively low during the study period, with fewer than one in every thirty customers making repeat purchases. Although the platform was successful in attracting new customers, it faced challenges in encouraging customers to return for subsequent purchases. Improving customer retention through loyalty programs, personalized recommendations, promotional offers, excellent customer service, and efficient delivery could help increase repeat purchases and enhance customer lifetime value.

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
#Applying of Smote

from imblearn.over_sampling import SMOTE

smote = SMOTE(random_state=42)

X_train_smote, y_train_smote = smote.fit_resample(
    X_train_scaled,
    y_train
)

print(y_train_smote.value_counts())
```

In this study, customer satisfaction was represented using the review score provided by customers after receiving their orders. The original review score ranges from 1 to 5. ### Feature Importance Analysis

After selecting the Random Forest model as the best-performing classifier, a feature importance analysis was conducted to identify the variables that contributed most to predicting customer review outcomes. Feature importance measures the relative contribution of each predictor to the model's decision-making process. Variables with higher importance scores have a greater influence on the model's predictions.

The analysis revealed that **payment value** was the most influential predictor, with an importance score of **0.1314**. This indicates that the total amount paid by customers played the greatest role in determining whether a customer would leave a positive or negative review. The **freight value** (0.1146) and **product price** (0.1108) were the second and third most important predictors, suggesting that financial aspects of the purchase significantly influence customer perceptions and review behavior.

Product-related characteristics also contributed substantially to the model's predictions. **Product weight** (0.0966) and **payment installments** (0.0893) ranked among the top five most important features. This suggests that both product attributes and payment flexibility influence customers' overall evaluation of their shopping experience.

Conversely, several seller location variables and a number of product category indicators contributed very little to the prediction of customer reviews. For example, variables such as **seller_state_PI**, **seller_state_PA**, **seller_state_AM**, and product categories including **fashion** and **security and services** recorded extremely low importance values, indicating that they had minimal influence on the model's predictive performance.

Overall, the feature importance analysis demonstrates that **transaction-related variables**, particularly payment value, freight value, product price, product weight, and payment installments, were the strongest determinants of customer review outcomes in the Olist e-commerce dataset. These findings suggest that customers place considerable emphasis on the financial cost of their purchases and the characteristics of the products received when evaluating their shopping experience. In contrast, geographic seller information and several product categories contributed relatively little to predicting customer reviews.

The results provide valuable insights for e-commerce businesses by highlighting the factors that should receive greater managerial attention. Strategies aimed at optimizing product pricing, reducing shipping costs, offering flexible payment options, and maintaining product quality are likely to improve customer experiences and increase the likelihood of receiving positive reviews.
